# Practicing Databricks platform

### Initialize
Select the required catalog.schema

```python
%sql
use catalog dbacademy;
use schema get_started_de;
select current_catalog(), current_schema();
```

Display all volumes

```python
%sql
show volumes;
```

Query the volumes/csvs

```python
%sql
select *  from read_files('/Volumes/dbacademy/get_started_de/myfiles/employees2.csv')
```

```python
%sql
LIST '/Volumes/dbacademy/get_started_de/myfiles/';
```

Create a table from the csv file/ Using CTAS to create tables

```python
%sql
Create table if not exists employees as select *  from read_files('/Volumes/dbacademy/get_started_de/myfiles/employees.csv')
```

```python
%sql
Create table if not exists current_employees_ctas as
select ID, FirstName,Role from read_files('/Volumes/dbacademy/get_started_de/myfiles/employees.csv',
format => 'csv',
header => true,
inferSchema => true );
```

### CRUD
View all data in my table

```python
%sql
select * from current_employees_ctas ;
```

Inserting data

```python
%sql
insert into employees (ID, FirstName, Country, Role) values
(7, "Shruti", "India", "Data Engineer");
```

```python
%sql
insert into employees select * from read_files('/Volumes/dbacademy/get_started_de/myfiles/employees2.csv')
```

Deleting data

```python
%sql
delete from employees where ID = 5 or ID = 6;
```

Updating data


```python
%sql
update employees set employees.FirstName = "Tanasha"  , employees.Country = "USA" where employees.ID = 5;
```

```python
%sql
update employees set employees.Role = "Data Engineer" where employees.ID = 7;
```

### Versioning
