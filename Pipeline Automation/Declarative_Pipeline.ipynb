```python
#selecting the catalog where our tables will exist
spark.sql("USE CATALOG adi_personal")
spark.sql("CREATE SCHEMA if not EXISTS databricks_lr")
spark.sql("Use schema databricks_lr")
```

```python
#creating the bronze table
spark.sql("""
          Create or refresh streaming table bronze_table_sdp
          Comment "Raw employee data ingested from CSV files in the myfiles volume."
          as select *
          from stream read_files(
            '/Volumes/dbacademy/get_started_de/myfiles',
            format => 'csv',
            header => true,
            inferSchema => true
          )""")

          %sql
```

```python
#creating the silver level view
create or refresh materialized view silver_view_sdp(
    constraint valid_id expect (ID is not null),
    constraint valid_name expect (FirstName is not null)
)
comment "Cleaned and enriched data with better quality expectations"
as select 
  ID,
  FirstName,
  Country,
  upper(Role) as Role,
  current_timestamp() as processed_timestamp,
  current_date() as processed_date
  from bronze_table_sdp


  %sql
```
