```python
#selecting the catalog where our tables will exist
spark.sql("USE CATALOG adi_personal")
spark.sql("CREATE SCHEMA if not EXISTS databricks_lr")
spark.sql("Use schema databricks_lr")
```

```python
#create bronze table and load data
spark.sql("""CREATE TABLE IF NOT EXISTS pipeline_bronze (
    ID INT,
    FirstName STRING,
    Country STRING,
    Role STRING
)""")

result = spark.sql(""" copy into pipeline_bronze 
                    from '/Volumes/dbacademy/get_started_de/myfiles'
                    FILEFORMAT = CSV
                    FORMAT_OPTIONS ('header' = 'true', 'inferSchema' = 'true')
                   """)

result.display()

print("Bronze table loaded")
```
