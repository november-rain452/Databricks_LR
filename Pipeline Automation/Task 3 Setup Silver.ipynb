```python
# set catalog and schema
spark.sql("USE CATALOG adi_personal")
spark.sql("USE SCHEMA databricks_lr")
```

```python
# create silver table
spark.sql("""Create or replace table pipeline_silver as
          select 
            ID,
            FirstName,
            Country,
            UPPER(Role) AS Role,
            current_timestamp() as processed_timestamp,
            current_date() as processed_date
           from pipeline_bronze 
            """)

print("Silver table created")
```
