```python
# set catalog and schema
spark.sql("USE CATALOG adi_personal")
spark.sql("USE SCHEMA databricks_lr")
```

```python
#create gold table 
spark.sql("""
          CREATE TABLE if not EXISTS pipeline_gold (
              Role STRING,
              TotalEmployees INT
          )
""")
```

```python
#insert data into gold table
spark.sql("""
          insert overwrite pipeline_gold
          select role, count(*) as Total_Employees
          from pipeline_silver
          group by role""")


print("Gold table created")
```
